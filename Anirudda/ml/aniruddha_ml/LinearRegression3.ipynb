A higher Adjusted R-squared value indicates that the regression model explains a greater proportion of the variance in the target variable, after accounting for the number of predictors used.

## OLS - Ordinary Least Square

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

!gdown 1UpLnYA48Vy_lGUMMLG-uQE1gf_Je12Lh
```

```python
df = pd.read_csv('cars24-car-price-clean.csv')
df.head()
```

```python
y=df[['selling_price']]
X=df.drop('selling_price', axis=1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
y_train = np.array(y_train)
```

```python
X_sm = sm.add_constant(X_train) # Statsmodel default is without intercept (wo), to add intercept use this code.

model_sm = sm.OLS(y_train, X_sm)
```

```python
model_sm
```

```python
results = model_sm.fit()
```

```python
print(results.summary())
```

```python
y_pred = results.predict(X_sm)
```
