```python
import tensorflow as tf
from tensorflow.keras.layers import Embedding
import numpy as np

# -----------------------------
# Sample sentence
# -----------------------------
sentence = ["I", "love", "transformers"]

# -----------------------------
# Create simple vocabulary
# -----------------------------
vocab = {
    "I": 1,
    "love": 2,
    "transformers": 3
}

# Convert words to token IDs
token_ids = [vocab[word] for word in sentence]

print("Token IDs:", token_ids)

# Output:
# [1, 2, 3]

# -----------------------------
# Parameters
# -----------------------------
max_length = 10
vocab_size = 100
embedding_dim = 8

# -----------------------------
# Token Embedding Layer
# -----------------------------
token_embedding = Embedding(
    input_dim=vocab_size,
    output_dim=embedding_dim
)

# -----------------------------
# Positional Embedding Layer
# -----------------------------
position_embedding = Embedding(
    input_dim=max_length,
    output_dim=embedding_dim
)

# -----------------------------
# Create position indices
# -----------------------------
positions = tf.range(start=0, limit=len(token_ids), delta=1)

print("Positions:", positions.numpy())

# Output:
# [0 1 2]

# -----------------------------
# Convert token IDs to tensor
# -----------------------------
token_tensor = tf.constant([token_ids])

# Shape: (1, sequence_length)

# -----------------------------
# Get embeddings
# -----------------------------
token_embeds = token_embedding(token_tensor)

position_embeds = position_embedding(positions)

# -----------------------------
# Add them together
# -----------------------------
final_embeddings = token_embeds + position_embeds

# -----------------------------
# Results
# -----------------------------
print("\nToken Embeddings Shape:")
print(token_embeds.shape)

print("\nPosition Embeddings Shape:")
print(position_embeds.shape)

print("\nFinal Embeddings Shape:")
print(final_embeddings.shape)

print("\nFinal Embeddings:")
print(final_embeddings.numpy())
```

```python
import tensorflow as tf
import numpy as np

# -----------------------------------
# Positional Encoding Function
# -----------------------------------
def positional_encoding(max_position, d_model):

    # Create matrix of shape:
    # (max_position, d_model)

    position = np.arange(max_position)[:, np.newaxis]

    # Shape: (max_position, 1)

    div_term = np.exp(
        np.arange(0, d_model, 2) *
        -(np.log(10000.0) / d_model)
    )

    # Shape: (d_model/2,)

    pe = np.zeros((max_position, d_model))

    # Apply sine to even indices
    pe[:, 0::2] = np.sin(position * div_term)

    # Apply cosine to odd indices
    pe[:, 1::2] = np.cos(position * div_term)

    # Add batch dimension
    pe = pe[np.newaxis, ...]

    return tf.cast(pe, dtype=tf.float32)


# -----------------------------------
# Example
# -----------------------------------

max_position = 10
d_model = 8

pos_encoding = positional_encoding(max_position, d_model)

print("Shape:")
print(pos_encoding.shape)

print("\nPositional Encoding Matrix:\n")
print(pos_encoding.numpy())
```
