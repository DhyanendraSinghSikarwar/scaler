Having explored measures of variability such as variance and standard deviation which particularly deal with random variables, let's try to understand what exactly a random variable is.

Random variables are essential for modelling and understanding uncertainty in various scenarios, let's have a look.

- Weather forecasts
- Stock market predictions
- Even a coin toss has uncertainty

- weather forecasts often express the probability of rain or storms, while in a coin toss, the probability of getting heads is quantified as 0.5


<br>

To mathematically account for such randomness, we use <font color='red'>**Random variables**</font>


A random variable is a situation/event/experiment, for which we are not certain about the outcome.

- <font color='orange'>Discrete RV</font>
- <font color='orange'>Continuous RV</font>

## Conditional Probability

### <font color='green'>**Experiment: Sum of 2 Dice Throws**</font>

Consider an experiment, where you are throwing 2 dices, and you are keeping a track of the sum of numbers obtained.

In order to represent the same, we create a table as shown. Here:
- Rows represent the number obtained from Dice 1
- Columns represent the number obtained from Dice 2
- Hence, the element corresponding to a given row and column represent the sum of numbers obtained on both dices (D1 + D2)

<br>

![image](https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/180/005/original/download_%281%29.png?1770893785)

Now using the table, you need to answer some questions.

> <font color='red'>**Q. What is the probability of getting a 2 on Dice 1?** </font>

Here, we do not care about the number obtained by Dice 2, we only care if Dice 1 yields a `2`.

Let's highlight the events of interest on the table. (Orange highlight)

So we know that this probability is $P(D_1 = 2) = \frac{6}{36}$

<br>

> <font color='red'> **Q. What is the probability that the sum of values obtained on D1 and D2 is less than or equal to 5?** </font>

Let's identify these events on the table. (Blue highlighted)

We have 10 such events, therefore the probability becomes: $P(D_1 + D_2 <= 5) = \frac{10}{36}$

<br>

> <font color='red'>**Q. What would be the probability of both these events happening together?** </font>

So, we want to find $P(D_1 = 2 ∩ D_1 + D_2 <= 5)$
- Here $∩$ means <font color='orange'>**Intersection**</font>

Note that there are only 3 such events (common part between orange and blue highlight).

So, probability becomes: $P(D_1 = 2 ∩ D_1 + D_2 <= 5) = \frac{3}{36}$

<br>

<font color='red'>Let's look at another question </font>

- Suppose an experiment is being conduted by 2 friends: Arun and Sapan
- Arun rolls both dices, but hides the result from Sapan
- When Sapan asks him what was observed, but Arun wants to mess with him instead
- So he tells him that the sum of both these nunbers is less than 5

Now he asks Sapan:

#### <font color='red'>**Q. What is the probability that $D_1 = 2$, given  $D_1 + D_2 <= 5$?**</font>

Before you try to answer this question, let's pause and try to understand what is being asked.

> <font color='red'>**What do we need to find?** </font>

To represent this mathematically, we will introduce a new terminology: $P(D_1 = 2 | D_1 + D_2 <=5)$



<br>

When we say that event $D_1 + D_2 <=5$ has already happened, our sample space is no longer 36.

It has now reduced to 10 possible outcomes (blue highlighted)

Within these 10, only 3 outcomes satisfy $D_1 = 2$ (pink highlighted).

Hence, the probability becomes: $P(D_1 = 2 | D_1 + D_2 <=5) = \frac{3}{10}$

<Br>

<font color='red'>**Note:** </font>
- The only difference between $P(D_1 = 2 \ ∩ \ D_1 + D_2 <=5)$ and $P(D_1 = 2 \ | \ D_1 + D_2 <=5)$ is that
 - The sample space of $P(D_1 = 2 \ ∩ \ D_1 + D_2 <=5)$ composes of all possible outcomes, i.e. 36
 - The sample space of $P(D_1 = 2 \ | \ D_1 + D_2 <=5)$ composes only the outcomes where $D_1 + D_2 <=5$ is found to be satisfied, i.e. 10
- This is known as <font color='orange'>**Conditional Probability** </font>

Let's formally define this.

The conditional probability $P(D_1 = 2 \ | \ D_1 + D_2 <=5)$ is equivalent to: <font color='red'> $\frac{P(D_1 = 2 \ ∩ \ D_1 + D_2 <=5)}{P(D_1 + D_2 <=5)}$ </font>

We can verify this: $P(D_1 = 2 \ | \ D_1 + D_2 <=5) = \frac{P(D_1 = 2 \ ∩ \ D_1 + D_2 <=5)}{P(D_1 + D_2 <=5)} = \frac{3/36}{10/36} = \frac{3}{10}$

![image](https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/180/012/original/download_%282%29.png?1770895603)

To summarize in general terms:
```
Probability of Event A, given Event B has already happened, is equivalent to the probability of A∩B ,
divided by probability of event B
```

i.e. <font color='red'>$P(A|B) = \frac{P(A∩B)}{P(B)}$ </font>

This equation is known as the <font color='orange'>**Conditional Probability Formula** </font>.

Let's analyse this further.

Multiplying both sides with $P(B)$, we get: <font color='red'>$P(A∩B) = P(A|B).P(B)$ </font>

In probability and stats, this is known as the <font color='orange'>**Product / Multiplication Rule**.</font>


<Br>

Similarly, we can expand $P(B∩A) = P(B|A).P(A)$


<font color='red'>Note: </font>
- This is not to be confused with the equation we saw in last class: $P(A∩B) = P(A).P(B)$
- Though it works in some problems, that equation is **not absolutely true**
- We will see when that is true, and when it is not through the concept of Dependent and Independent events later in next lecture.
- However, this Multiplication Rule is **absolutely true**. It is valid in ALL situations.

## Marginal and Joint probabilities

Till now we've solved quite a few problems. It's time to formally define the different types of probability values that we've been using.


We are given a table containing data about Sachin's match history.

> <font color='red'>**Do you recall from last class, how can we get such a table from a given dataframe?** </font>

We can use the function `pandas.crosstab()` with `margins = True`

<Br>

Let's define the events happening here:
- $W$: Sachin's team winning the match
- $C$: Sachin scoring a century

<br>

Let's answer a few questions based on this table

> <font color='red'>**What is the probability that Sachin's team wins the match?** </font>

We need to find $P(W) = \frac{No \ of \ matches \ won \ by
 \ Sachin}{Total  \ no \ of \ matches} = \frac{184}{360}$

><font color='red'> **What is the probability of Sachin scoring a century?** </font>

$P(C) = \frac{No \ of \ matches \ with \ century}{Total  \ no \ of \ matches} = \frac{46}{360}$

<br>

Similarly, we can calculate $P(W^C)$ and $P(C^C)$ as well.

All of these probability values are known as <font color='orange'>**Marginal Probability** </font>
- It is the probability of an event irrespective of the outcome of other variable.
- For instance, consider $P(W)$
 - It denotes the total probability of Sachin's team winning the match, considering both possibilities that Sachin may or may not score a century.
- It is not conditioned on another event. It may be thought of as an **unconditional probability.**
- Other example:  
 - Probability that a card drawn is a 4  : P(four)=1/13.
   - This includes the possibility of the 4 being a spades, heart, club or diamond.
 - Probability that a card drawn is spades  : P(spades)=1/4.


Now let's look at the second type of probability values, by answering the following questions.

> <font color='red'>**What is the probability that Sachin's team wins AND he scores a century?** </font>

We need to find $P(W∩C) = \frac{30}{360}$

<br>

Similarly, we can find $P(C∩W)$, $P(W^C \cap C^C)$ and $P(W \cap C^C)$



<Br>

- Here we calculated the likelihood of two events **occurring simultaneously**, essentially the **intersection of the two events** only. This kind of probability is known as <font color='orange'>**Joint Probability**.</font>

  - Represented as $P(A∩B)$

- Other Example:  the probability that a card is a four and red = P(four and red) = 2/52

![image](https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/180/014/original/download_%284%29.png?1770895966)

The third kind of probability value, we've just studied, i.e. <font color='orange'>**Conditional Probability**. </font>

Let's answer a few questions on this also

> <font color='red'>**What is the probability that Sachin's team wins the match given that he scored a century?** </font>

Since it is given that he scores a century, our subset reduces to the second row.

Now since we want to find the prob of team winning among these matches, our probability becomes: $P(W|C) = \frac{30}{46}$

> <font color='red'>**What is the probability that Sachin scores a century, given that his team has won the match?** </font>

As per the given extra information, our subset reduces to the second column.

So among these 184 matches, where India won, Sachin scored a century in only 30 matches.

Therefore $P(C|W) = \frac{30}{184}$

<br>

Similarly, you can be asked to calculate other conditional probabilities such as: $P(W|C^C)$, $P(W^C|C)$, $P(C|W^C)$, etc.

![image](https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/180/015/original/download_%285%29.png?1770896120)

## T-test

### <font color='red'>**Motivation:**</font>

<font color='red'>**Let's say you are a Research Scientist working on a new cognitive enhancement pill**</font>

- The goal is to develop a pill that can significantly improve IQ scores in individuals.

- The researchers believe that the new pill will lead to a significant increase in average IQ scores for the population.


<Br>

### <font color='red'>Use Case: Improve IQ with Pill</font>

Suppose that the average IQ of the population is known to be 100

A researcher claims that his pill will improve IQ

> <font color='red'>**What is the first thing that you will do?**</font>

Collect data/evidence, and then try to test if his hypothesis is correct.

````python
# The pill is given to a few people and their IQ is tested with following results:
iq_scores = [110, 105, 98, 102, 99, 104, 115, 95]
````



<font color='red'>**Testing the Pill's Effectiveness:**</font>

**Case 1:** The new pill may increase the average IQ scores.

- By enhancing cognitive functions, individuals taking the pill may show improved performance in IQ tests.

**Case 2:** The new pill may have no significant effect on IQ scores.

- The researchers need to verify whether the pill is effective or not before widespread use.

> <font color='red'>**What are the null and alternate hypothesis?**</font>

- $H_0: μ = 100$ (pill has no effect)
- $H_a: μ > 100$ (pill has positive effect)

<br>
<br>

<font color='red'>**Testing the Hypothesis:**</font>

- Researchers need a way to test whether the new pill has a significant impact on IQ scores.
- This involves comparing the average IQ scores of a group taking the pill with the population average.



Let's see the mean IQ of this sample

````python
np.mean(iq_scores)
````

> <font color='red'>The mean seems to be $> 100$, can we directly say that okay this pill is effective?</font>

No. We'd have to conduct a hypothesis test and then we can validate this claim with confidence.

<br>

> <font color="red">How should we proceed? Can we apply Z Test here?</font>

**Scenario Complexity:**
- Note that here, we have no clue about the value of population std deviation.
- Hence, we would not be able to calcluate the Z-test statistic.
- In fact, in most real world scenarios, the **population standard deviation is often unknown.**

**Sample Size:**
- Also note that the sample size here is very small (< 30).
- We do not have enough data.

> <font color='red'>**Can we plot the distribution?**</font>

- No.
- We know that the sample mean is 100
- But we do not know the sample standard deviation (std error) or population standard deviation.

<br>

Moreover, if we had that, we would've easily computed the z-score corresponding to $x = 103.5$ and then the pvalue

But this is not possible here. We **cannot use z-score / z-statistic**

> <font color='red'>**What to do now?**</font>

We do not have a population standard deviation, so we cannot evaluate it for the sample, as per CLT.

However, we are given the sample, so we can compute the **sample standard deviation ($s$).**

When we do that, we will get: $\frac{x - μ}{\frac{s}{\sqrt{n}}}$

- $n$: sample size
- $s$: sample standard deviation

This is **NOT the z-score anymore**, in fact, it is known as the **T-Statistic** that yields **T Distribution**

This modified framework is called **T Test**.


When dealing with small sample sizes, the use of the t-test is more appropriate as it **accounts for the increased uncertainty associated with smaller samples.**



Depending on the situation, there are 3 variations of T-tests:-
- With 1 sample
 - Here, You have a bunch of samples that you are comparing with a single number
- With 2 samples where samples are independent
- With 2 samples where samples are dependent

In our problem, we are given only
- Population mean: 100 (a single number)
- One set of sample IQ values: `[110, 105, 98, 102, 99, 104, 115, 95]`

Hence, it is a case of **One Sample T Test**

<br>



In this framework of T-test, we no longer need to use `.cdf(), .pdf(), etc`.

We will use a one-line code, where we directly use the existing function from scipy.

We need to call it: `ttest_1samp(sample, 100)`
- Here, 100 is the number we want to compare, as per the null hypothesis.

As a result, it will give us:
- T statistic value
- p-value

## Co-variance matrix


In machine learning, we rarely deal with
single numbers. We deal with massive
vectors-images, sensor arrays, or financial
ledgers-where dimensions ($d$) can
number in the thousands.

To make sense of a
vector $x$ ϵ $R^d$, we cannot look at features in
isolation. We must understand their
geometry: how they spread, how they
cluster, and how they define a 'shape' in space.



![image](https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/180/122/original/img1.png?1770971226)

Variance measures the spread of a
single variable. Covariance measures
how two variables, $X_i$ and $X_j$, move
together.

When we summarise these
relationships for all pairs of variables,
we get the Covariance Matrix, V.

- it is an $n × n$ matrix where the
$(i, j)_{th}$ entry is Cov [$X_i, X_j$].
- It is always symmetric and positive
semi-definite that is, $z^TVz ≥ 0$).

Processing high-dimensional data is
expensive and noisy. We want to map
data points $x_i$ ϵ $R^d$ to a lower-
dimensional feature space $z_i$ ϵ $R^k$
without losing the essential structure.

**The solution:** Project the data onto a
unit vector $u$.
