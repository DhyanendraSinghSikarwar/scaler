### 1. **Question:**
A line in the plane is written as $2x - ky + 7 = 0$. You want it to be **parallel** to the line $4x - 6y + 1 = 0$. Which value of $k$ ensures parallelism?

- A) $k = -3$
- B) $k = 3$
- C) $k = \frac{6}{4} = 1.5$
- D) $k = \frac{6}{2} = 3$

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:**   D) $k = \frac{6}{2} = 3$

**Explanation:**  
Two lines $A_1 x + B_1 y + C_1 = 0$ and $A_2 x + B_2 y + C_2 = 0$ are parallel if their ratios $A_1/B_1 = A_2/B_2$. Here, compare $2 / (-k)$ with $4 / (-6)$. Thus, $\frac{2}{-k} = \frac{4}{-6}$ $\implies$ $\frac{2}{k} = \frac{4}{6} = \frac{2}{3}$. Solve for $k=3$.

First, slope of $L_1$ from $(2,3)$ to $(4,7)$ is $\frac{7 -3}{4 -2}=\frac{4}{2}=2.$ A line perpendicular to slope $2$ must have slope $-\frac{1}{2}$ (negative reciprocal).

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:**   A) Subtract the lines’ distances from origin

**Explanation:**

In the context of Principal Component Analysis (PCA), the term "rank-deficient" refers to a situation where the covariance matrix, denoted as Σ, has eigenvalues that are equal to zero.

The covariance matrix Σ is a square matrix of shape (d x d), where d represents the number of dimensions or features in the dataset. The eigenvalues of Σ represent the variance explained by each principal component.

It's important to note that a rank-deficient covariance matrix can lead to a loss of information and potentially affect the performance of subsequent data analysis tasks that rely on PCA, such as dimensionality reduction or feature selection.
