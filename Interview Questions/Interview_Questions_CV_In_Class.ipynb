#### 1. **Why does a convolution layer with *kernel size = 1* never change an image’s spatial dimensions, no matter the stride or padding?**

* **A)** Because the kernel has no spatial extent, so padding is implicitly disabled.
* **B)** Because each output position corresponds to exactly one input position, leaving height and width untouched.
* **C)** Because deep-learning libraries automatically replace $1 \times 1$ kernels with fully connected layers.
* **D)** Because stride is forced to 1 internally whenever the kernel is $1 \times 1$.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **B)** Each output activation is computed from the same spatial coordinate in the input, so size remains unchanged regardless of other settings.

**Explanation:** A $1 \times 1$ kernel behaves like a per-pixel linear mixer over channels. Stride or padding can still be set, but they do not alter spatial resolution because the receptive field is confined to one location.

* **C)** “Valid” means no bias term is used; “same” includes a bias.
* **D)** “Valid” converts convolution into cross-correlation; “same” keeps the mathematical flip.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **B)** “Valid” avoids padding, shrinking the map, while “same” pads to preserve spatial dimensions.

**Explanation:** The choice affects output size, not kernel mathematics, dilation, or bias handling.

#### 4. **Increasing a convolution’s *kernel size* while holding everything else fixed most directly impacts which property?**

* **A)** The number of learnable parameters grows quadratically with kernel side length.

<br>
<br>
<br>

**Correct Answer:** **A)** Each step from $k$ to $k+1$ in both dimensions adds an entire ring of weights, inflating parameters as $k^2$.

**Explanation:** Nothing in the definition of convolution automatically scales stride, rescales activations, or breaks differentiability.

#### 1. **Why does *spatial pooling* (e.g., max-pooling) often improve a CNN’s ability to generalize to unseen images?**

* **A)** Pooling explicitly encodes rotation invariance by averaging out orientation information across channels.
* **B)** Pooling reduces the number of learnable parameters, decreasing reliance on precise spatial positions of features while retaining their presence.
* **C)** Pooling enforces a strict one-to-one correspondence between every input pixel and exactly one output activation, preventing over-compression.
* **D)** Pooling eliminates high-frequency components, guaranteeing that low-frequency (texture-based) features dominate the representation, which is always desirable.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **B)** Pooling reduces the number of learnable parameters, decreasing reliance on precise spatial positions of features while retaining their presence.

**Explanation:** By summarizing activations within local neighborhoods, pooling layers shrink the feature map and detach feature **existence** from **exact location**. Fewer parameters (and fewer activations flowing into deeper layers) lower the risk of memorizing pixel-specific patterns, helping the network generalize to translations and modest distortions. Pooling does not directly encode rotation invariance (A), enforce one-to-one mappings (C), nor guarantee that suppressing high-frequency detail is always optimal (D).

#### 2. **In transfer learning for vision, why does *fine-tuning only the last few layers* of a deep pretrained CNN often work surprisingly well on a new but related dataset?**

* **A)** Early layers capture generic low-level features (edges, textures) that are broadly useful, so retaining them lets the model focus capacity on adapting high-level representations.
* **B)** Freezing most layers forces the network to retrain from scratch, preventing catastrophic forgetting of the original dataset.
* **C)** Later layers inherently encode dataset-specific color statistics, so adjusting them alone corrects mismatched illumination between datasets.
* **D)** Fine-tuning only the final layers reduces training time but invariably leads to poorer accuracy compared to full-network training.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **A)** Early layers capture generic low-level features (edges, textures) that are broadly useful, so retaining them lets the model focus capacity on adapting high-level representations.

**Explanation:** Convolutional hierarchies transition from universal local patterns to dataset-specific semantics. Keeping early layers fixed leverages their reusable features while updating the task-specific top layers, often achieving strong accuracy with limited data and compute. It *doesn’t* retrain from scratch (B), color statistics are dispersed across layers (C), and performance can match or exceed full fine-tuning when data are scarce (D).

#### 3. **Batch normalization is ubiquitous in deep vision models. From a first-principles viewpoint, which statement best captures its primary contribution during *inference* (test time)?**

* **A)** It regularizes by injecting random noise through mini-batch statistics, enhancing robustness to domain shift.
* **B)** It linearly rescales activations to a narrower dynamic range learned during training, stabilizing subsequent nonlinearities.
* **C)** It adaptively drops out units whose activations exceed a learned threshold, preventing feature co-adaptation.
* **D)** It propagates gradient signals more effectively by re-centering activations around zero, which is critical only during back-propagation.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **B)** It linearly rescales activations to a narrower dynamic range learned during training, stabilizing subsequent nonlinearities.

**Explanation:** At inference, batch-norm uses *fixed* running-mean and running-variance estimates to shift and scale activations, ensuring they sit in a range where the following nonlinearities behave predictably. No random noise is injected (A), no units are dropped (C), and gradient flow is irrelevant at inference (D).

#### 4. **Why do modern object-detection pipelines (e.g., Faster R-CNN, YOLOv8) apply *non-maximum suppression* (NMS) after generating raw bounding-box predictions?**

* **A)** NMS prunes detections with low center-point confidence, ensuring class probabilities sum to one across all boxes.
* **B)** NMS enforces diverse aspect ratios by discarding boxes whose shape closely matches the dataset average.
* **C)** NMS removes highly overlapping boxes with similar scores, consolidating multiple detections of the same object into a single prediction.
* **D)** NMS balances class imbalance by oversampling boxes from under-represented classes and discarding others.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **C)** NMS removes highly overlapping boxes with similar scores, consolidating multiple detections of the same object into a single prediction.

**Explanation:** Dense detectors produce many overlapping proposals around the same object. NMS keeps the highest-scoring box and suppresses nearby ones whose *intersection-over-union* exceeds a threshold, yielding one box per object. It does **not** enforce probability normalization (A), shape diversity (B), or class re-balancing via oversampling (D).

#### 5. **Why do *residual connections* (as in ResNet) make it feasible to train CNNs hundreds of layers deep?**

* **A)** They drastically reduce the total parameter count by re-using convolution kernels across multiple layers.
* **B)** They create shortcut paths that let gradients flow around nonlinearities, alleviating vanishing-gradient problems in very deep stacks.
* **C)** They enforce orthogonality among filters, preventing correlation between feature maps.
* **D)** They randomly drop entire feature maps during training, mimicking an aggressive form of dropout.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **B)** They create shortcut paths that let gradients flow around nonlinearities, alleviating vanishing-gradient problems in very deep stacks.

**Explanation:** The identity skip lets gradients back-propagate almost unattenuated, so very deep networks learn without collapsing. Skip-connected blocks neither reuse weights (A), nor impose orthogonality (C), nor act as dropout (D).

#### 9. **When applying a $3 \times 3$ convolution with *stride 2* and *no padding* to an input feature map, what is the most direct conceptual reason the output’s spatial resolution drops?**

<br>
<br>

<br>
<br>

**Correct Answer:** **B)** The kernel “skips” every other input location, so fewer distinct positions are evaluated.

#### 10. **In a standard 2-D convolution layer, why does increasing the *number of output channels* (filters) enlarge the parameter count even though kernel size and stride stay fixed?**

* **A)** Each new output channel has its *own* set of weights spanning *all* input channels, so parameters grow linearly with the number of filters.
* **B)** Output channels are created by duplicating and renormalizing existing filters, doubling parameters geometrically.
* **C)** Adding filters triggers automatic use of dilated kernels, which insert trainable “holes.”
* **D)** More filters require larger padding, introducing extra learnable border weights.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **A)** Each new output channel has its own weight tensor (kernel) connected to every input channel, so adding filters increases parameters linearly.

**Explanation:** For a $(k \times k)$ kernel with $C\_{\text{in}}$ input channels, each additional filter contributes $k \times k \times C\_{\text{in}}$ new weights (plus bias). Options B–D are not how frameworks implement channel expansion.

<br>
<br>
<br>
<br>
<br>
<br>

**Correct Answer:** **B)** Linearity means partial derivatives accumulate as sums of products—exactly what a convolution does—so frameworks reuse the same primitive.
