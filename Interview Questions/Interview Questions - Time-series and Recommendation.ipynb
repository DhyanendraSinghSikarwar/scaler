# Interview Questions

### 1. **Question:**
You have a daily demand series showing no clear seasonality but a gradual upward drift. If you decide to use a “naive” forecast  rather than a “drift” forecast, what might be a likely outcome?

- A) The naive forecast will effectively capture the upward trend
- B) The naive forecast could lead to persistent underestimation or overestimation as the series drifts upward
- C) The naive forecast provides the same result as a rolling average
- D) The naive forecast adjusts automatically for drift over time

<br>
<br>
<br>
<br>
<br>
<br>
**Correct Answer:** B) The naive forecast could lead to persistent underestimation or overestimation as the series drifts upward

**Explanation:** A naive forecast replicates the last data point, ignoring any gradual drift. Over time, if a series has a positive (or negative) trend, naive forecasts will lag behind (or lead) the actual values.


### 2. **Question:**
A monthly series with strong holiday effects might be well-served by a “seasonal naive” model. Which scenario best describes when a seasonal naive forecast might fail?

- A) When the series has a different pattern each year, even for the same months
- B) When the series exhibits stable and recurring seasonal behavior
- C) When the series has zero seasonality
- D) When the series is purely random with no structured signals

<br>
<br>
<br>
<br>
<br>
<br>
**Correct Answer:** A) When the series has a different pattern each year, even for the same months

**Explanation:** Seasonal naive repeats the value from the same period in the previous season. If seasonal patterns aren’t consistent year-over-year (i.e., the underlying seasonality changes), this method can produce poor forecasts.

### 46. **Question:**
A rule `Bagel ⇒ Cream Cheese` yields:

- Support(`Bagel` & `Cream Cheese`) = 0.20  
- Confidence(`Bagel ⇒ Cream Cheese`) = 0.80

\[

= \frac{0.20}{0.125}
= 1.6
\]  
$$

A lift above 1 indicates a positive association between bagels and cream cheese relative to chance.
